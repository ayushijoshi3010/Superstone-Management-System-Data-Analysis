# Data Type Conversion

```python
# Convert Date columns into date Datatype

df['Order Date'] = pd.to_datetime(
    df['Order Date'],
    format='mixed',
    dayfirst=True
)

df['Ship Date'] = pd.to_datetime(
    df['Ship Date'],
    format='mixed',
    dayfirst=True
)

# Check Conversions

print("\nDate Columns Converted\n")

print(f"Order Date: {df['Order Date'].dtype}")
print(f"Ship Date: {df['Ship Date'].dtype}")

df.head(5)


```

# UNIVARIATE ANALYSIS

```python
print("==CATEGORICAL COLUMNS==")
categorical_cols=['Customer Segment','Category','Region','Payment Mode','Delivery Status','Auto Reorder']

for col in categorical_cols:
    print(f"\n*--{col}--*")
    print(f"\n{df[col].value_counts()}")

# Visualization

    plt.figure(figsize=(6,6))
    df[col].value_counts().plot(kind='bar',color='pink')
    plt.title(f'Distribution of {col}')
    plt.xticks(rotation=30)
    plt.tight_layout()
    plt.show()
```

# Numerical Variable Analysis

```python
print("\n=== NUMERICAL VARIABLES ===\n")

numerical_cols=['Quantity','Unit Price','Discount %','Sales Amount','Cost Price','Profit','Stock Left','Reorder Quantity']

for col in numerical_cols:
    print(f"\n*---{col}---*\n")
    print(f"Mean : {df[col].mean():.2f}")
    print(f"Median : {df[col].median():.2f}")
    print(f"Standard Deviation : {df[col].std():.2f}")
    print(f"Min : {df[col].min():.2f}")
    print(f"Max : {df[col].max():.2f}")

    # Visulaization

    plt.figure(figsize=(8,4))

    plt.subplot(1,2,1) # Histogram for distribution of Data
    df[col].hist(bins=30,color='lightgreen',alpha=0.7)
    plt.title(f'Distribution of {col}')

    plt.subplot(1,2,2) #Boxplot to check if any outliers present in data or not.
    df.boxplot(column=col)
    plt.title(f"Boxplot of {col}")

    plt.tight_layout()
    plt.show()
```

# BIVARIATE ANALYSIS

```python
print("\n== SALES & PROFIT BY CATEGORY ==\n")
category_performance=df.groupby('Category').agg({
    'Sales Amount':'sum',
    'Profit':'sum',
    'Order Id':'count'
}).rename(columns={'Order Id':'Order Count'})

# Creating new column in category performance i.e profit margin..

category_performance['Profit Margin']=(category_performance['Profit']/category_performance['Sales Amount'])*100

print(category_performance)

# Visualization

fig,axes=plt.subplots(2,2,figsize=(15,12))
# Sales By Category
category_performance['Sales Amount'].plot(kind='bar',ax=axes[0,0],color='blue',alpha=0.7)
axes[0,0].set_title("Sales By Category")
axes[0,0].tick_params(axis='x',rotation=30)

# Profit By Category
category_performance['Profit'].plot(kind='bar',ax=axes[0,1],color='blue',alpha=0.7)
axes[0,1].set_title("Profit By Category")
axes[0,1].tick_params(axis='x',rotation=30)

# Order Count by Category
category_performance['Order Count'].plot(kind='bar',ax=axes[1,0],color='blue',alpha=0.7)
axes[1,0].set_title("Order Count By Category")
axes[1,0].tick_params(axis='x',rotation=30)

# Profit Margin by Category
category_performance['Profit Margin'].plot(kind='bar',ax=axes[1,1],color='blue',alpha=0.7)
axes[1,1].set_title("Profit Margin By Category")
axes[1,1].tick_params(axis='x',rotation=30)

plt.tight_layout()
plt.show()
```

# Regional Performance

```python
regional_performance=df.groupby('Region').agg({
    "Profit":"sum",
    "Sales Amount":"sum",
    "Order Id":"count",
}).rename(columns={"Order Id":"Order Count"})
regional_performance['Profit Margin']=(regional_performance['Profit']/regional_performance['Sales Amount'])*100
print(f"\n=== Regional Performance === \n{regional_performance}")

fig,axes=plt.subplots(2,2,figsize=(16,8))
regional_performance['Profit'].plot(kind='pie',autopct='%1.1f%%',ax=axes[0,0])
axes[0,0].set_ylabel("")
axes[0,0].set_title("Profit By Region")


regional_performance['Sales Amount'].plot(kind='pie',autopct='%1.1f%%',ax=axes[0,1])
axes[0,1].set_ylabel("")
axes[0,1].set_title("Sales By Region")

regional_performance['Order Count'].plot(kind='pie',autopct='%1.1f%%',ax=axes[1,0])
axes[1,0].set_ylabel("")
axes[1,0].set_title("Order By Region")

regional_performance['Profit Margin'].plot(kind='pie',autopct='%1.1f%%',ax=axes[1,1])
axes[1,1].set_ylabel("")
axes[1,1].set_title("Profit Margin By Region")

plt.tight_layout()
plt.show()
```

# MONTHLY TRENDS

```python
print("\n== Monthly Trends ==\n")

monthly_trends = df.groupby(['Order Year','Order Month']).agg({
    'Profit': 'sum',
    'Sales Amount': 'sum',
    'Order Id': 'count'
}).rename(columns={'Order Id': 'Order Count'})


# Profit Margin
monthly_trends['Profit Margin'] = (
    monthly_trends['Profit'] / monthly_trends['Sales Amount']
) * 100


# # Correct month order
# month_order = [
#     'January', 'February', 'March', 'April',
#     'May', 'June', 'July', 'August',
#     'September', 'October', 'November', 'December'
# ]

# monthly_trends = monthly_trends.reindex(month_order)

print(monthly_trends)

# Visualization

figure, axes = plt.subplots(4, 1, figsize=(16, 18))


# 1. Monthly Profit
monthly_trends['Profit'].plot(
    kind='line',
    color='green',
    ax=axes[0],
    alpha=0.7,
    marker='o'
)

axes[0].set_title("Monthly Profit")
axes[0].set_xlabel("Order Month")
axes[0].set_ylabel("Profit")
axes[0].ticklabel_format(style='plain', axis='y')
axes[0].grid(True)


# 2. Monthly Sales
monthly_trends['Sales Amount'].plot(
    kind='line',
    color='green',
    ax=axes[1],
    alpha=0.7,
    marker='o'
)

axes[1].set_title("Monthly Sales")
axes[1].set_xlabel("Order Month")
axes[1].set_ylabel("Sales Amount")
axes[1].ticklabel_format(style='plain', axis='y')
axes[1].grid(True)


# 3. Monthly Orders
monthly_trends['Order Count'].plot(
    kind='line',
    color='green',
    ax=axes[2],
    alpha=0.7,
    marker='o'
)

axes[2].set_title("Monthly Orders")
axes[2].set_xlabel("Order Month")
axes[2].set_ylabel("Order Count")
axes[2].grid(True)


# 4. Monthly Profit Margin
monthly_trends['Profit Margin'].plot(
    kind='line',
    color='green',
    ax=axes[3],
    alpha=0.7,
    marker='o'
)

axes[3].set_title("Monthly Profit Margin")
axes[3].set_xlabel("Order Month")
axes[3].set_ylabel("Profit Margin (%)")
axes[3].grid(True)


plt.tight_layout()
plt.show()
```

# ADVANCED ANALYSIS

### 1)Correlation

```python
print("\n===CORRELATION ANALYSIS===\n")

numerical_df=df[['Quantity','Unit Price','Discount %','Sales Amount','Cost Price','Profit']]
numerical_df

correlation_matrix=numerical_df.corr()
print(correlation_matrix)

plt.figure(figsize=(10,8))
sns.heatmap(correlation_matrix,
            annot=True,
            cmap='coolwarm',
            center=0
           )

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

# Customer Segment Analysis

```python
print('\n===Segment Analysis===\n')
segment_analysis=df.groupby('Customer Segment').agg({
    'Sales Amount':['sum','mean'],
    'Profit':['sum','mean'],
    'Order Id':'count',
    'Discount %':'mean'
        
}).round(2)
# print(segment_analysis)
segment_analysis.columns=['Total Sales','Avg Sales','Total Profit','Avg Profit','Order Count','Avg Discount']
print(segment_analysis)
# Visualization

fig, axes=plt.subplots(2,2,figsize=(15,10))

segment_analysis['Total Sales'].plot(kind='bar',ax=axes[0,0],color='lightblue')
axes[0,0].set_title("Total Sales by Customer Segment")
axes[0,0].tick_params(axis='x',rotation=30)

segment_analysis['Total Profit'].plot(kind='bar',ax=axes[0,1],color='lightgreen')
axes[0,1].set_title("Total Profit by Customer Segment")
axes[0,1].tick_params(axis='x',rotation=30)

segment_analysis['Avg Profit'].plot(kind='bar',ax=axes[1,0],color='pink')
axes[1,0].set_title("Avg Profit by Customer Segment")
axes[1,0].tick_params(axis='x',rotation=30)

segment_analysis['Avg Discount'].plot(kind='bar',ax=axes[1,1],color='lightblue')
axes[1,1].set_title("Avg Discount by Customer Segment")
axes[1,1].tick_params(axis='x',rotation=30)

plt.tight_layout()
plt.show()
```

# KEY INSIGHTS AND SUMMARY

```python
# Summarizing everything in one place now.

print("\n*=== KEY INSIGHTS SUMMARY ===*\n")

# Overall Mtrics

total_sales=df['Sales Amount'].sum()
total_profit=df['Profit'].sum()
total_orders=df['Order Id'].nunique()
avg_profit_margin=(total_profit/total_sales)*100


print(f"\n**OVERALL PERFORMANCE**\n")
print(f"Total Sales ₹{total_sales:,.2f}")
print(f"Total Profit ₹{total_profit:,.2f}")
print(f"Total Orders {total_orders}")
print(f"Average Profit Margin ₹{avg_profit_margin:,.2f}")


# Top Performing Categoreis
top_categories=df.groupby('Category')['Profit'].sum().sort_values(ascending=False)
# print(f"\n\n*****Top Categories By Profit******\n\n{top_categories}")


print("\n\n**Profit Made By Categories**")
for i, (category,profit) in enumerate(top_categories.items(),1):
    print(f"{i}. {category} : {profit:,.2f}")
    

# Regional Performance
best_region=df.groupby('Region')['Profit'].sum().idxmax()
print(f"\n**BEST PERFORMING REGION **: {best_region}")


# Delivery Status
print("\n\n**DELIVERY STATUS**")
best_delivery=df['Delivery Status'].value_counts(normalize=True)*100
for status,percentage in best_delivery.items():
    print(f"{status} : {percentage:,.2f}%")

    
# Customer Segment
best_segment=df.groupby('Customer Segment')['Profit'].sum().idxmax()
print(f"\nMOST PROFITABLE CUSTOMER SEGMENT: {best_segment}")
```

# EXPORT RESULTS

```python
analysis_results={
    'total_sales':total_sales,
    'total_profit':total_profit,
    'total_orders':total_orders,
    'avg_profit_margin':avg_profit_margin,
    'top_category':top_categories.index[0],
    'best_region':best_region,
    'best_customer_segment':best_segment
}
# Convert to Data Frame 
results_df=pd.DataFrame([analysis_results])
results_df

results_df.to_csv('superstone_eda_analyis_summary.csv',index=False)
print("\nAnalysis summary saved to 'eda_analysis_summary.csv'")
```
